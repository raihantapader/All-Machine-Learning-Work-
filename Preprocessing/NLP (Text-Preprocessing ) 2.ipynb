```python
## importing the dataset
import pandas as pd

messages = pd.read_csv("SMSSpamCollection", sep="\t", names=["Output","Input"])
messages
```

```python
messages.shape        ## shape is used for to see the total number of inputs
```

```python
## Data cleaning and preprocessing
import re  ## re means regular expression
import nltk ## nltk means natural languale toolkit
nltk.download('stopwords')
```

```python
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer  ## its use for Stemming purpose
ps=PorterStemmer()
```

```python
import re
```

```python
stopwords_set = set(stopwords.words('english'))
```

```python
## Apply Stopwords. Lemmatize

corpus = []
for i in range(len(messages)):  # Correctly using 'len(sentences)' to determine the number of iterations
    review = re.sub('[^a-zA-Z]', ' ', messages['Input'][i])  # Accessing each sentence using 'sentences[i]'
    review = review.lower() ## convert lower case and duplicate word not come.
    review = review.split() ##that basically means i'm going to get each and every senteces
    
    review = [ps.stem(word) for word in review if not word in stopwords_set]
    
    ## review = [lemmatizer.lemmatize(word) for word in review if not word in set(stopwords.words('english'))]
    review = ' '.join(review)
    corpus.append(review)
```

```python
corpus
```

```python
## Creating the Bag of words model
from sklearn.feature_extraction.text import CountVectorizer ## CountVectorizer is basically used for to create BOW over there
cv= CountVectorizer(max_features=2000,binary=True,ngram_range=(2,2))
X=cv.fit_transform(corpus).toarray()
```

```python
X.shape
```

```python
y=pd.get_dummies(messages['Output'])
y=y.iloc[:,1].values
```

```python
y
```

```python
## Train Test Split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test= train_test_split(X,y,test_size=0.20,random_state=0)
```

```python
X_train,y_train         ## input and output features
```

```python
from sklearn.naive_bayes import MultinomialNB
spam_detect_model=MultinomialNB().fit(X_train,y_train)
```

```python
## Prediction
y_pred=spam_detect_model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,classification_report
```

```python
score=accuracy_score(y_test,y_pred)
print(score)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_pred,y_test))
```
