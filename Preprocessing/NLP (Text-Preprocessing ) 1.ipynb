```python

corpus = []
for i in range(len(sentences)):  # Correctly using 'len(sentences)' to determine the number of iterations
    review = re.sub('[^a-zA-Z]', ' ', sentences[i])  # Accessing each sentence using 'sentences[i]... Removing all special characters lilke capital A - Z '
    review = review.lower()
    corpus.append(review)
```

```python
nltk.download('stopwords')
```

```python
stopwords.words('english')  # Define the `stopwords` variable correctly
```

```python
##stemming

# Iterate over each sentence in the corpus
for i in corpus:
    words = nltk.word_tokenize(i)  # Tokenize the sentence into words
    for word in words:
        if word not in set(stopwords.words('english')):  # Check if the word is not a stopword
            print(stemmer.stem(word))  # Stem the word and print it

            
```

```python
## Lemmatization
for i in corpus:
    words=nltk.word_tokenize(i)
    for word in words:
        if word not in set(stopwords.words('english')):
            print(lemmatizer.lemmatize(word))
```

```python
## Apply Stopwords. Lemmatize
import re
corpus = []
for i in range(len(sentences)):  # Correctly using 'len(sentences)' to determine the number of iterations
    review = re.sub('[^a-zA-Z]', ' ', sentences[i])  # Accessing each sentence using 'sentences[i]'
    review = review.lower()
    review = review.split() ##that basically means i'm going to get each and every senteces
    review = [lemmatizer.lemmatize(word) for word in review if not word in set(stopwords.words('english'))]
    review = ' '.join(review)
    corpus.append(review)
```
