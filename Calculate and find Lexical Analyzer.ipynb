```python
statement = input("Enter a Statement: ")

statement = statement.split()

keyword_list = ['if', 'printf', 'for', 'int', 'float', 'double', 'auto', 'break', 'else', 'long', 'signed', 'goto', 'then', 'while','char','void','return']

identifier = []
keyword = []
operator = []
arithmetic_operator = []
relational_operator = []
assignment_operator = []
bitwise_operator = []
puncuation = []
integers = []
floating = []

for i in range(len(statement)):
    #operator check
    if statement[i] == '+' or statement[i] == '-' or statement[i] == '*' or statement[i] == '/' or statement[i] == '%' or statement[i] == '=' or statement[i] == '&' or statement[i] == '|' or statement[i] == '<' or statement[i] == '>':
        operator.append(statement[i])

    #relational operator check
    if statement[i] == '==' or statement[i] == '>' or statement[i] == '<' or statement[i] == '!=' or statement[i] == '>=' or statement[i] == '<=':
        relational_operator.append(statement[i])

    #assignment operator check
    if statement[i] == '=':
        assignment_operator.append(statement[i])

    #bitwise operator check
    if statement[i] == '&' or statement[i] == '|' or statement[i] == '~':
        bitwise_operator.append(statement[i])

    #puncuation check
    if statement[i] == ';' or statement[i] == ',' or statement[i] == ':' or statement[i] == '(' or statement[i] == ')' or statement[i] == '{' or statement[i] == '}' or statement[i] == '[' or statement[i] == ']':
        puncuation.append(statement[i])

    #keyword check
    if statement[i] in keyword_list:
        keyword.append(statement[i])

    #integers check
    if statement[i].isnumeric():

        integers.append(statement[i])

    #floats check
    try:
        float(statement[i])
        floating.append(statement[i])
    except ValueError:
        pass



print("OPERATOR: ")
print ("Arithmetic operator: ", operator)
print ("Relational operator: ", relational_operator)
print ("Assignment operator: ", assignment_operator)
print ("Bitwise operator: ", bitwise_operator)
print ("PUNCTION SYMBOL: ", puncuation)
print ("Keywords: ", keyword)
print ("Integers: ", integers)
print ("Floats: ", floating)
```

```python
def is_keyword(word):
    keywords = ['if', 'then', 'else', 'for', 'while', 'int', 'float', 'char', 'return', 'void']
    return word in keywords

def is_operator(ch):
    operators = ['+', '-', '*', '/', '%', '=', '&', '|', '<', '>']
    return ch in operators

def print_token_map(token_map, token_type):
    print(f"{token_type}:")
    for token, count in token_map.items():
        print(f"{token} ({count})", end=" ")
    print()


statement = input("Enter a statement: ")

keywords_map = {}
identifiers_map = {}
arith_operators_map = {}
assign_operators_map = {}
rel_operators_map = {}
bitwise_operators_map = {}
punctuation_map = {}
int_map = {}
float_map = {}

buffer = ""
i = 0

while i < len(statement):
    if statement[i].isalnum() or statement[i] == '.':
        buffer += statement[i]
    else:
        if buffer:
            if is_keyword(buffer):
                keywords_map[buffer] = keywords_map.get(buffer, 0) + 1
            elif buffer[0].isdigit():
                if '.' in buffer:
                    float_map[buffer] = float_map.get(buffer, 0) + 1
                else:
                    int_map[buffer] = int_map.get(buffer, 0) + 1
            else:
                identifiers_map[buffer] = identifiers_map.get(buffer, 0) + 1
            buffer = ""

        if is_operator(statement[i]):
            op = statement[i]
            if statement[i] == '=' and i + 1 < len(statement) and statement[i + 1] == '=':
                op += '='
                rel_operators_map[op] = rel_operators_map.get(op, 0) + 1
                i += 1
            elif statement[i] in ['&', '|']:
                bitwise_operators_map[op] = bitwise_operators_map.get(op, 0) + 1
            elif statement[i] == '=':
                assign_operators_map[op] = assign_operators_map.get(op, 0) + 1
            else:
                arith_operators_map[op] = arith_operators_map.get(op, 0) + 1
        elif statement[i] in [';', ',', '(', ')', '{', '}', '[', ']', ':']:
            punctuation_map[statement[i]] = punctuation_map.get(statement[i], 0) + 1
    i += 1

# Token Output
print("\nToken Output:\n")
print_token_map(keywords_map, "KEYWORDS")
print_token_map(identifiers_map, "IDENTIFIERS")
print("OPERATORS:")
print_token_map(arith_operators_map, "\tArithmetic operators")
print_token_map(rel_operators_map, "\tRelational operators")
print_token_map(assign_operators_map, "\tAssignment operators")
print_token_map(bitwise_operators_map, "\tBitwise operators")
print_token_map(int_map, "\tINTEGERS")
print_token_map(float_map, "\tFLOATS")
print_token_map(punctuation_map, "PUNCTUATION SYMBOLS")
```
